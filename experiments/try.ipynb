```python
# Compare blending vs prefix-only results (including latency)
import pandas as pd
import numpy as np

blend_path = '/home/gcp-vm/projects/LMCache/experiments/script/results_blending.csv'
prefix_path = '/home/gcp-vm/projects/LMCache/experiments/script/results_prefix_only.csv'

df_blend = pd.read_csv(blend_path)
df_prefix = pd.read_csv(prefix_path)

print('Loaded:')
print(f'  blending rows:    {len(df_blend)}')
print(f'  prefix_only rows: {len(df_prefix)}')

# Keep core columns if present
core_cols = [
    'request_name',
    'chunk_token_length',
    'elapsed_seconds',
    'lmcache_num_hit_tokens',
    'lmcache_num_lookup_hits',
    'lmcache_num_vllm_hit_tokens',
    'hit_chunk_hashes_count',
    'new_chunk_hashes_count',
    'segment_hit_hashes_count',
    'segment_new_hashes_count',
]

avail_blend = [c for c in core_cols if c in df_blend.columns]
avail_prefix = [c for c in core_cols if c in df_prefix.columns]

blend = df_blend[avail_blend].copy()
prefix = df_prefix[avail_prefix].copy()

# Aggregate by request/chunk for stable comparison
keys = [k for k in ['request_name', 'chunk_token_length'] if k in blend.columns and k in prefix.columns]
if not keys:
    keys = ['request_name']

blend_g = blend.groupby(keys, dropna=False).mean(numeric_only=True).reset_index()
prefix_g = prefix.groupby(keys, dropna=False).mean(numeric_only=True).reset_index()

cmp = prefix_g.merge(blend_g, on=keys, how='outer', suffixes=('_prefix', '_blend'))

# Latency deltas and speedup
if 'elapsed_seconds_prefix' in cmp.columns and 'elapsed_seconds_blend' in cmp.columns:
    cmp['latency_delta_s (blend-prefix)'] = cmp['elapsed_seconds_blend'] - cmp['elapsed_seconds_prefix']
    cmp['latency_speedup_x (prefix/blend)'] = cmp['elapsed_seconds_prefix'] / cmp['elapsed_seconds_blend']

# Useful metric deltas
for m in [
    'lmcache_num_hit_tokens',
    'lmcache_num_lookup_hits',
    'lmcache_num_vllm_hit_tokens',
    'hit_chunk_hashes_count',
    'segment_hit_hashes_count',
]:
    p = f'{m}_prefix'
    b = f'{m}_blend'
    if p in cmp.columns and b in cmp.columns:
        cmp[f'{m}_delta (blend-prefix)'] = cmp[b] - cmp[p]

# Overall latency summary
print('=== Overall Latency Summary ===')
if 'elapsed_seconds' in prefix.columns and 'elapsed_seconds' in blend.columns:
    p_mean = prefix['elapsed_seconds'].mean()
    b_mean = blend['elapsed_seconds'].mean()
    print(f'prefix mean latency:  {p_mean:.6f}s')
    print(f'blend  mean latency:  {b_mean:.6f}s')
    print(f'delta (blend-prefix): {b_mean - p_mean:+.6f}s')
    if b_mean > 0:
        print(f'speedup (prefix/blend): {p_mean / b_mean:.4f}x')

print('=== Per-Request Comparison ===')
display_cols = keys + [
    c for c in [
        'elapsed_seconds_prefix',
        'elapsed_seconds_blend',
        'latency_delta_s (blend-prefix)',
        'latency_speedup_x (prefix/blend)',
        'lmcache_num_hit_tokens_prefix',
        'lmcache_num_hit_tokens_blend',
        'lmcache_num_hit_tokens_delta (blend-prefix)',
        'lmcache_num_lookup_hits_prefix',
        'lmcache_num_lookup_hits_blend',
        'lmcache_num_lookup_hits_delta (blend-prefix)',
        'lmcache_num_vllm_hit_tokens_prefix',
        'lmcache_num_vllm_hit_tokens_blend',
        'lmcache_num_vllm_hit_tokens_delta (blend-prefix)',
        'hit_chunk_hashes_count_prefix',
        'hit_chunk_hashes_count_blend',
        'hit_chunk_hashes_count_delta (blend-prefix)',
        'segment_hit_hashes_count_prefix',
        'segment_hit_hashes_count_blend',
        'segment_hit_hashes_count_delta (blend-prefix)',
    ] if c in cmp.columns
]

cmp = cmp.sort_values(keys).reset_index(drop=True)
display(cmp[display_cols])

```
